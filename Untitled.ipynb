# 🚦 Traffic Sign Classification Project Report

## 1. Introduction
Traffic signs play a vital role in ensuring road safety by providing critical information to drivers. They regulate traffic flow, warn of potential hazards, and guide drivers to make safe decisions. For autonomous vehicles and advanced driver assistance systems (ADAS), recognizing traffic signs accurately is essential to mimic human driving behavior and comply with traffic laws.

The **German Traffic Sign Recognition Benchmark (GTSRB)** dataset is a widely used benchmark for evaluating traffic sign classification models. It contains images of 43 different traffic sign classes captured under varying conditions such as illumination, occlusion, rotation, and perspective distortion. These challenges make the dataset an excellent testbed for computer vision and deep learning techniques.

In this project, we aim to:
- Preprocess and analyze the GTSRB dataset.
- Build a Convolutional Neural Network (CNN) model using Keras/TensorFlow.
- Train and evaluate the model on traffic sign images.
- Achieve high accuracy suitable for real-world applications like autonomous driving.
- Visualize results and discuss potential improvements.

This project demonstrates the application of deep learning in solving a real-world problem and highlights the importance of preprocessing, model design, and evaluation in building robust AI systems.


## 2. Dataset Description
- **Source:** GTSRB dataset (Meta, Train, Test folders with CSV metadata).
- **Classes:** 43 traffic sign categories.
- **Images:** ~50,000 images with varying sizes, lighting, and angles.
- **Preprocessing:**
  - Resized all images to **32×32 pixels**.
  - Normalized pixel values to range [0,1].
  - Labels converted to **one-hot encoding**.

We also visualize random samples from the dataset and plot the class distribution to understand data balance.

```python
IMG_SIZE = 32
data = []
labels = []

# Path to Meta folder and CSV file
dataset_path = r"Meta/"
csv_file = r"Meta.csv"

df = pd.read_csv(csv_file)

for i, row in df.iterrows():
    img_path = os.path.join(dataset_path, row['Path'])
    img = cv2.imread(img_path)
    if img is None:
        continue
    img = cv2.resize(img, (IMG_SIZE, IMG_SIZE))
    data.append(img)
    labels.append(row['ClassId'])  # class ID from CSV

data = np.array(data) / 255.0
labels = np.array(labels)
```

```python

IMG_SIZE = 32

# Paths
train_csv = r"Train.csv"
test_csv = r"Test.csv"

# Load Train Data
df_train = pd.read_csv(train_csv)
X_train, y_train = [], []

for i, row in df_train.iterrows():
    img_path = row['Path']   # <-- use Path column
    img = cv2.imread(img_path)
    if img is None:
        continue
    img = cv2.resize(img, (IMG_SIZE, IMG_SIZE))
    X_train.append(img)
    y_train.append(row['ClassId'])

X_train = np.array(X_train) / 255.0
y_train = to_categorical(np.array(y_train), 43)

# Split into train/validation
X_train, X_val, y_train, y_val = train_test_split(
    X_train, y_train, test_size=0.2, random_state=42
)

# Load Test Data
df_test = pd.read_csv(test_csv)
X_test, y_test = [], []

for i, row in df_test.iterrows():
    img_path = row['Path']
    img = cv2.imread(img_path)
    if img is None:
        continue
    img = cv2.resize(img, (IMG_SIZE, IMG_SIZE))
    X_test.append(img)
    y_test.append(row['ClassId'])

X_test = np.array(X_test) / 255.0
y_test = to_categorical(np.array(y_test), 43)
```

```python
# Show 10 random samples from training data
plt.figure(figsize=(15,5))
for i in range(10):
    idx = np.random.randint(0, len(X_train))
    plt.subplot(2,5,i+1)
    plt.imshow(X_train[idx])
    plt.title(np.argmax(y_train[idx]))  # show class ID
    plt.axis('off')
plt.suptitle("Sample Traffic Sign Images", fontsize=16)
plt.show()
```

## 3. Methodology
### Preprocessing
- Loaded images using OpenCV (`cv2.imread`).
- Resized to uniform dimensions.
- Normalized pixel intensities.
- Split dataset into **train (80%)**, **validation (20%)**, and separate **test set**.

### Model Architecture (CNN in Keras)
- **Conv2D layers:** Extract spatial features.
- **MaxPooling2D:** Reduce dimensionality.
- **BatchNormalization + Dropout:** Improve generalization and prevent overfitting.
- **Dense layers:** Fully connected for classification.
- **Softmax output:** 43 neurons for multi-class classification.


```python
model = Sequential()

model.add(Conv2D(32,(3,3), activation='relu', input_shape=(32,32,3)))
model.add(BatchNormalization())
model.add(MaxPooling2D(pool_size=(2,2)))

model.add(Conv2D(64,(3,3), activation='relu'))
model.add(BatchNormalization())
model.add(MaxPooling2D(pool_size=(2,2)))
    
model.add(Conv2D(128,(3,3), activation='relu'))
model.add(BatchNormalization())
model.add(MaxPooling2D(pool_size=(2,2)))

model.add(Flatten())
model.add(Dense(128, activation='relu'))
model.add(Dropout(0.5))
model.add(Dense(43,activation='softmax'))
```

```python
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
```

```python
model.summary()
```

## 4. Results
- **Training Accuracy:** ~99%
- **Validation Accuracy:** ~95%
- **Test Accuracy:** ~96%

### Performance Visualizations
- Accuracy and loss curves over epochs.
- Confusion matrix showing misclassifications.


```python
history = model.fit(X_train, y_train, batch_size=64, epochs=20, validation_data=(X_val, y_val))
```

```python
test_loss, test_acc = model.evaluate(X_test, y_test)
print("Test Accuracy:- ", test_acc)
```

```python
plt.plot(history.history['accuracy'], label = 'train acc')
plt.plot(history.history['val_accuracy'], label = 'val acc')
plt.legend()
plt.show()
```

```python
y_pred = model.predict(X_test)
y_pred_classes = np.argmax(y_pred, axis=1)
y_true = np.argmax(y_test, axis=1)
```

```python
cm = confusion_matrix(y_true, y_pred_classes)
sns.heatmap(cm, annot=False, cmap='Blues')
plt.show()
```
